```python
import numpy as np
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.optim as optim
import torch.nn.init as init
import torch.nn.functional as F

print("all complete")
```

```python
# 데이터 정의
x_data = [[80, 220, 6300],
          [75, 167, 4500],
          [86, 210, 7500],
          [110, 330, 9000],
          [95, 280, 8700],
          [67, 190, 6800],
          [79, 210, 5000],
          [98, 250, 7200]]

y_data = [2, 3, 1, 0, 0, 3, 2, 1]  # 분류 레이블 (0~3)
```

```python
# 텐서 변환
x_train = torch.FloatTensor(x_data)
y_train = torch.LongTensor(y_data)  # 레이블은 정수형(Long)이어야 함
```

```python
# 데이터 모양 확인
print(x_train.shape)
print(y_train.shape)
```

```python
# 가중치(W)와 편향(b) 초기화
# 입력 특성이 3개, 출력 클래스가 4개이므로 (3, 4) 크기
W = torch.zeros((3, 4), requires_grad=True)
b = torch.zeros(1, requires_grad=True)

# 최적화 도구 설정 (확률적 경사 하강법)
optimizer = optim.SGD([W, b], lr=0.00001)
```

```python
W1 = torch.zeros((3, 4), requires_grad=True)
b1 = torch.zeros(1, requires_grad=True) 
optimizer1 = optim.SGD([W1, b1], lr=0.00001)
costs1 = [] 
```

```python
for epoch in range(10000):
    optimizer1.zero_grad()
    z = torch.sigmoid(x_train.matmul(W1) + b1)
    cost1 = F.cross_entropy(z, y_train)
    cost1.backward()
    optimizer1.step()
    costs1.append(cost1.item())
    if epoch % 10 == 0:
        print('epoch:%d cost:%f' % (epoch, cost1.item()))
```

```python
# 가중치(W)와 편향(b) 초기화
# 입력 특성이 3개, 출력 클래스가 4개이므로 (3, 4) 크기
W = torch.zeros((3, 4), requires_grad=True)
b = torch.zeros(1, requires_grad=True)
```

```python
# sigmoid 는 이진분류를 위해 사용했지만 현재 다중 분류에는 맞지 않는다고 생각해서 기본 w_1+b_1 으로 변경
for epoch in range(10000):
    optimizer1.zero_grad()
    z = x_train.matmul(W1) + b1 
    cost1 = F.cross_entropy(z, y_train)
    cost1.backward()
    optimizer1.step()
    costs1.append(cost1.item())
    if epoch % 10 == 0:
        print('epoch:%d cost:%f' % (epoch, cost1.item()))
```

```python
# 가중치(W)와 편향(b) 초기화
# 입력 특성이 3개, 출력 클래스가 4개이므로 (3, 4) 크기
W = torch.zeros((3, 4), requires_grad=True)
b = torch.zeros(1, requires_grad=True)
```

```python
# 정규화 문제라고 판단 
x_norm = (x_train - x_train.mean(dim=0)) / x_train.std(dim=0)
```

```python
for epoch in range(10000):
    optimizer1.zero_grad()
    
    z = x_norm.matmul(W1) + b1 
    
    cost1 = F.cross_entropy(z, y_train)
    cost1.backward()
    optimizer1.step()
    costs1.append(cost1.item())
    
    if epoch % 10 == 0:
        print('epoch:%d cost:%f' % (epoch, cost1.item()))
```

```python
optimizer = optim.SGD([W, b], lr=0.1)
for epoch in range(10000):
    optimizer.zero_grad()
    
    z = x_norm.matmul(W1) + b1 
    
    cost1 = F.cross_entropy(z, y_train)
    cost1.backward()
    optimizer1.step()
    costs1.append(cost1.item())
    
    if epoch % 1000 == 0:
        print('epoch:%d cost:%f' % (epoch, cost1.item()))
```
